```python
import itertools
import numpy as np
from operator import itemgetter

N = 8
hs = [5, -3, 7, 0, -2, 4, 6, 1]
Js = [-3, 0, -2, -5, -3, 2, -4]

# The cost function encoded by the Ising Hamiltonian
def cost_function(s): 
    return -s @ np.diag(Js, k=1) @ s - np.dot(hs, s)

# Compute the minimum by checking all possible solutions
gse, gs = min(((cost_function(np.array(s)), s) for s in itertools.product(*[(-1, 1)] * N)), key=itemgetter(0))

print("Ground State: ", gs)
print("Ground State Energy: ", gse)
```
